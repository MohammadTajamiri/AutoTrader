```python
import numpy as np
import pandas as pd
```

```python
bodyType = pd.read_csv("BodyType.csv")
Enrichment = pd.read_csv("DataEnrichment.csv")
```

```python
bodyType["Make_Model"]=(bodyType["Make"]+" " + bodyType["Model"]+" " + bodyType["MoreInformation"].fillna("")).str.split()
bodyType["Quality Check"] =(bodyType["Make"]+" " + bodyType["Model"]+" " + bodyType["MoreInformation"].fillna("")).str.split()
```

```python
bodyType["Make_Model"] = bodyType["Make_Model"].apply(lambda x: " ".join(x) if isinstance(x, list) else x)
bodyType["Quality Check"] = bodyType["Quality Check"].apply(lambda x: " ".join(x) if isinstance(x, list) else x)
```

```python
Enrichment["Make_Model"]=Enrichment["Make"] + " " + Enrichment["Model"]
```

```python
df=Enrichment.merge(bodyType,on=["Make_Model"],how="left")
df.to_csv("MainDataEnrichement.csv")
```
